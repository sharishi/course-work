```python
def find_route(courier_location, orders_location, city_map_list):
    def is_valid_move(x, y):
        return 0 <= x < 100 and 0 <= y < 100 and city_map_list[y][x] == 1

    def calculate_distance(pos1, pos2):
        return abs(pos1[0] - pos2[0]) + abs(pos1[1] - pos2[1])

    def move_courier(current_location, destination):
        route = []
        x, y = current_location
        dest_x, dest_y = destination

        while (x, y) != destination:
            if dest_x > x and is_valid_move(x + 1, y):
                x += 1
            elif dest_x < x and is_valid_move(x - 1, y):
                x -= 1
            elif dest_y > y and is_valid_move(x, y + 1):
                y += 1
            elif dest_y < y and is_valid_move(x, y - 1):
                y -= 1

            route.append((x, y))

        return route

    route = [courier_location]
    current_location = courier_location

    # Сортировка заказов по расстоянию от текущего местоположения курьера
    sorted_orders = sorted(orders_location, key=lambda x: calculate_distance(current_location, x))

    for order in sorted_orders:
        route += move_courier(current_location, order)
        current_location = order

    return route

route = find_route(courier_location, orders_location, city_map_list)
```

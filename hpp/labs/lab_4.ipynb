Напишите код, который обрабатывает значение, введенное пользователем, и
выводит какое-то сообщение ему, используя оператор if...elif...else.

```python
def polite_recognizer(text):
    if text.startswith(('Hello', 'Hi')):
        return 'Your politeness level is 50%'
    elif text.startswith(('Good morning', 'Good night', 'Good evening', 'Good night')):
        return 'Your politeness level is 100%'
    else:
        return 'Your politeness level is 0%'
    
input_text = input("Write a sample address and we'll assess your level of politeness")
polite_recognizer(input_text)
```

Напишите код, который позволит создание счетчика, используя в этих целях цикл
while.

```python
import random
count = 0
random_limit = random.randint(1, 20)
while count < random_limit:
    print("Current counter:", count)
    count += 1
```

```python
http_status = 400
match http_status:
    case 400:
        print("Bad Request")
    case 403:
        print("Forbidden")
    case 404:
        print("Not Found")
    case _:
        print("Other")
```

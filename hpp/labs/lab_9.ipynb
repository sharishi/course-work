```python
import time
from datetime import datetime


start_time = time.time()

print("Запуск задачи...")
time.sleep(4)  

end_time = time.time()

duration = end_time - start_time
print(f"Задача завершена за {duration:.2f} секунд.")

# Представление времени в формате даты и времени
start_datetime = datetime.fromtimestamp(start_time)
end_datetime = datetime.fromtimestamp(end_time)

print("Начало:", start_datetime.strftime("%Y-%m-%d %H:%M:%S"))
print("Конец:", end_datetime.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
import calendar
from datetime import datetime

# Параметры для календаря
year = 2024
month = 11

# Выводим календарь на месяц
print(calendar.month(year, month))


cal = calendar.monthcalendar(year, month)

print("Дни месяца разделенные на рабочие и будние:")
for week in cal:
    for day in week:
        if day != 0:  # Проверяем, что день является числом (не пустой день)
            day_date = datetime(year, month, day)
            weekday = day_date.weekday()  # 0 - понедельник, 6 - воскресенье
            
            # Если день будний (с понедельника по пятницу), выводим его
            if weekday < 5:
                print(day_date.strftime("%Y-%m-%d"), "- Рабочий день")
            else:
                print(day_date.strftime("%Y-%m-%d"), "- Будний день")
```

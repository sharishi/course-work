## Задание 1 Создание списка покупок (shopping list)
a) Определите 3 функции: для добавления товара в списке, для удаления выбранного
товара из списка и для вывода всех элементов списка.
b) Создайте меню из 4-х опций:
1 Вывести список текущих товаров,
2 Добавить товар в списке, (на основе input()),
3 Удалить товар из списка,
4 Выход.
c) Создайте цикл, в котором пользователь выбирает опцию – а вы вызываете какую-то
функцию, на основе его выбора. Выход из цикла – когда пользователь вводит
опцию соответствующей «Выходу».

Пункт а

```python
def add_tolist(mylist, item):
    if item in mylist:
        user_input = input("This item is already in the list. Do you want to add it again?(Yes/No): ")
        if user_input.lower() == "no":
            return
    print(f"You just added {item} to the list")
    mylist.append(item)
def delete_fromlist(mylist, item):
    if item not in mylist:
        print("We don't have this item in our list. Please try again")
        return
    while item in mylist:
        mylist.remove(item)
    print(f"You just removed {item} from the list")
def print_list(my_list):
    print("Your list:")
    for i in my_list:
        print(i)
```

Пункт b

```python
def show_menu():
    print("Menu:")
    print("1. Display the list of current products: ")
    print("2. Add item to list: ")
    print("3. Remove an item from the list:")
    print("4. Exit:")
```

```python
def shop_list(mylist = []):
    while True:
        try:
            show_menu()
            user_input = int(input("Enter the option number: "))

            match user_input:
                case 1:
                    print_list(mylist)
                case 2:
                    item_input = str(input("Enter the item you want to add: "))
                    add_tolist(mylist, item_input)
                    print_list(mylist)
                case 3:
                    item_remove = str(input("Enter the item you want to remove: "))
                    delete_fromlist(mylist, item_remove)
                    print_list(mylist)
                case 4:
                    print("We'll be happy to see you again!")
                    break
                case _:
                    print("You entered the wrong option number. Please try again =)")
        except ValueError:
            print("Ошибка: неверный формат числа! Пожалуйста, введите число.")


mylist = ['turmeric', 'coffee', 'tea', 'croissant']
shop_list(mylist)
```

## Списки

```python
my_list = [1, 2, 3]

my_list.append(4)
print(f"После append(4): {my_list}")  # После append(4): [1, 2, 3, 4]

my_list.extend([4, 5])
print(f"После extend([4, 5]): {my_list}")  # После extend([4, 5]): [1, 2, 3, 4, 5]

my_list.insert(1, 10)
print(f"После insert(1, 10): {my_list}")  # После insert(1, 10): [1, 10, 2, 3, 4, 5]

item = my_list.pop()
print(f"После pop(): элемент {item}, список {my_list}")  # После pop(): элемент 5, список [1, 10, 2, 3, 4]


my_list.sort()
print(f"После sort(): {my_list}")  # После sort(): [1, 2, 3, 4, 10]
```

## Множества

```python
my_set = {1, 2, 3}

my_set.add(4)
print(f"После add(4): {my_set}")  # После add(4): {1, 2, 3, 4}

my_set.remove(2)
print(f"После remove(2): {my_set}")  # После remove(2): {1, 3, 4}

my_set.discard(5)  # Ошибки нет, если элемента нет
print(f"После discard(5): {my_set}")  # После discard(5): {1, 3, 4}

set1 = {1, 2, 3}
set2 = {3, 4, 5}
result_union = set1.union(set2)
print(f"После union(set2): {result_union}")  # После union(set2): {1, 2, 3, 4, 5}

result_intersection = set1.intersection(set2)
print(f"После intersection(set2): {result_intersection}")  # После intersection(set2): {3}
```

## Словари

```python
# Словари
my_dict = {'apple': 10, 'banana': 15}

value = my_dict.get('apple')
print(f"После get('apple'): {value}")  # После get('apple'): 10

keys = my_dict.keys()
print(f"После keys(): {keys}")  # После keys(): dict_keys(['apple', 'banana'])

values = my_dict.values()
print(f"После values(): {values}")  # После values(): dict_values([10, 15])

items = my_dict.items()
print(f"После items(): {items}")  # После items(): dict_items([('apple', 10), ('banana', 15)])

popped_value = my_dict.pop('apple')
print(f"После pop('apple'): {popped_value}, словарь: {my_dict}")  
# После pop('apple'): 10, словарь: {'banana': 15}
```

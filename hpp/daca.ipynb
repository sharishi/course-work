```python
def palindrome(string_pal):
    string_pal = string_pal.lower()
    return string_pal == string_pal[::-1]
```

```python
palindrome("Nin")
```

```python
def text_ver(doc):
    with open(doc, 'r') as file:
        text = file.read()
        
    words = text.split()

    for word in words:
        if 'a' in word:
            print(word)    
```

```python
my_file = 'try_w_a.txt'
text_ver(my_file)
```

```python
class Text:
    def __init__(self, string_pal, doc):
        self.string_pal = string_pal
        self.doc = doc
        
    def palindrome(self):
        return self.string_pal.lower() == self.string_pal[::-1]
    
    def text_ver(self):
        with open(self.doc, 'r') as file:
            text = file.read()
        
        words = text.split()

        for word in words:
           if 'a' in word:
               print(word) 
    

text = Text('nin', my_file)
print(text.palindrome())
text.text_ver()
```
